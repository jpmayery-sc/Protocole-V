Notebook de test pour Colab — simulation simple.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 500)
y = np.sin(x) * np.exp(-x*0.1)

plt.plot(x, y)
plt.title('Oscillation amortie')
plt.xlabel('x')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```
